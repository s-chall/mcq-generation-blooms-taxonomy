# Chi-square analysis data check

The publication's chi-square test compared human usability ratings with the number of item-writing-guideline passes. The required human-evaluation fields are not public, so this notebook verifies that limitation instead of fabricating results.

```python
from pathlib import Path
import csv

repo_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
data_path = repo_root / 'data' / 'demo_questions.csv'
with data_path.open(newline='', encoding='utf-8') as handle:
    fieldnames = tuple(csv.DictReader(handle).fieldnames or ())

required_for_publication_test = {'usability'}
missing = required_for_publication_test - set(fieldnames)
assert missing == {'usability'}
print('Chi-square test not run: public data does not include human usability ratings.')
```
